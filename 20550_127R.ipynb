```r
data <- data.frame(
  year = c(2019, 2020, 2021, 2022, 2023),
  count = c(5, 7, 6, 9, 11)
)

print(data)
```

```r
model <- lm(count ~ year, data = data)
summary(model)
```

```r
library(ggplot2)

ggplot(data, aes(x = year, y = count)) +
  geom_point() +
  geom_smooth(method = "lm", se = FALSE) +
  labs(title = "선형 회귀 분석",
       x = "연도",
       y = "개수") +
  theme_minimal()
```
